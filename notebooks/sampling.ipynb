```python
import pandas as pd
import os

def create_random_sample(output_filepath, fraction=0.05):
    """
    Loads a large CSV, takes a purely random sample, and saves it to a new file.
    """
    print(f"Loading massive dataset")

    DATA_PATH = "../data/raw/TEP_Faulty_Testing.csv" 

    df_faulty = pd.read_csv(DATA_PATH)
    df_normal = pd.read_csv("../data/raw/TEP_FaultFree_Testing.csv")
    df = pd.concat([df_normal, df_faulty], ignore_index=True)
    print(f"Taking a random {fraction * 100}% sample...")
    sampled_df = df.sample(frac=fraction, random_state=42)
    output_dir = os.path.dirname(output_filepath)
    if output_dir:
        os.makedirs(output_dir, exist_ok=True)
        
    # Save the new, smaller dataset without the index column
    sampled_df.to_csv(output_filepath, index=False)
    print(f"Success! Saved {len(sampled_df)} rows to {output_filepath}")

# ==========================================
# Execution
# ==========================================
output_file = '../data/raw/TEP_Random_Testing_Sample_5_percent.csv'

# Run it to grab 5% of the data
create_random_sample( 
    output_filepath=output_file, 
    fraction=0.05
)
```
